```python
#Add two numbers - Take two numbers from the user and print their sum
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
sum = a + b
print("Sum:" , sum)
```

```python
#Calculate Rectangle area - Take the length and width from the user and calculate the area
length = int(input("Enter length: "))
width = int(input("Enter width: "))
area = length * width
print("Area:" , area)
```

```python
# Convert Celsius to Fahrenheit- take temperature in Celsius and convert into Fahrenheit
Celsius = float(input("Enter temperatue in celsius:"))
fahrenhiet = (Celsius * 9/5) + 32
print("Temperature in fahrenhiet:" , fahrenhiet)
```

```python
# Simple interest- take principle, rate and time from the user and calculate simple interest
principle = float(input("Enter principle amount:"))
rate = float(input("Enter rate:"))
time = float(input("Enter time:"))
simple_interest = (principle * rate * time)/100
print("Simple interest:" , simple_interest)
```

```python
# Swap two numbers - take two numbers from the user and swap their values
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
print("Before Swapping:")
print("a=", a)
print("b=", b)
a,b = b,a
print("After Swapping:")
print("a=", a)
print("b=", b)
```

```python
# Calculate total and average marks - take marks of three Three subjects from the user and calculate the total Marks and average marks
marks1 = float(input("Enter marks of first subject: "))
marks2 = float(input("Enter marks of second subject: "))
marks3 = float(input("Enter marks of third subject: "))
total = marks1 + marks2 + marks3
average = total/3
print("Total marks:" , total)
print("Average marks:", average)
```

```python
# Shopping bill- take a price and quantity of an item calculate the total price, 10% discount, final amount
price = float(input("Enter price: "))
quantity =float(input("Enter quantity: "))
total = price * quantity
discount = total * 10/100
final_amount = total - discount
print("Total amount:" , total)
print("Discount:" , discount)
print("Final amount:" , final_amount)
```
